# Task 6: Bayesian Estimation

## Given information

The observations are modelled as a random sample from a gamma distribution with shape 3 and scale $1/\theta$.

The generated prior parameters are

$$
\alpha_0=45,\qquad \beta_0=35,
$$

where $\beta_0$ is interpreted as a scale parameter.

The observed sample mean is

$$
\overline{x}=64.22.
$$

The assignment initially specifies $x_1,\ldots,x_{10}$.
Therefore, the numerical calculation uses

$$
n=10.
$$

The subsequent use of n is treated as general sample-size notation.

## Objectives

1. Derive the posterior distribution of $\theta$
2. Find the Bayes point estimate under squared-error loss
3. Find the estimate given by the posterior mode

## Gamma parameterization and sampling model

The shape-scale gamma density is defined as

$$
g(u;\alpha,\beta)
=
\frac{u^{\alpha-1}e^{-u/\beta}}
{\Gamma(\alpha)\beta^\alpha},
\qquad u>0,
$$

with shape $\alpha>0$ and scale $\beta>0$.

Its rate parameter is the reciprocal of its scale.

Conditional on $\theta$, assume that the observations are independent and identically distributed:

$$
X_i\mid\theta
\sim
\operatorname{Gamma}
\left(\text{shape}=3,\ \text{scale}=\frac{1}{\theta}\right),
\qquad \theta>0.
$$

Substituting into the gamma density gives

$$
f(x_i\mid\theta)
=
\frac{\theta^3}{\Gamma(3)}
x_i^2e^{-\theta x_i},
\qquad x_i>0.
$$

Thus, $\theta$ acts as the rate of the sampling distribution.

The prior is

$$
\theta
\sim
\operatorname{Gamma}
\left(\text{shape}=\alpha_0=45,\ \text{scale}=\beta_0=35\right).
$$

Its density is

$$
\pi(\theta)
=
\frac{\theta^{\alpha_0-1}e^{-\theta/\beta_0}}
{\Gamma(\alpha_0)\beta_0^{\alpha_0}},
\qquad\theta>0.
$$

The prior rate is therefore $1/\beta_0=1/35$, not 35.

```python
from fractions import Fraction

sample_shape = 3
n = 10

# rep of the supplied sample mean
sample_mean_exact = Fraction("64.22")
sample_sum_exact = n * sample_mean_exact

# parameters: shape and SCALE
prior_shape = 45
prior_scale_exact = Fraction(35, 1)
prior_rate_exact = 1 / prior_scale_exact

assert n > 0
assert sample_shape > 0
assert prior_shape > 0
assert prior_scale_exact > 0

print(f"Sample size: {n}")
print(f"Sample mean: {float(sample_mean_exact):.2f}")
print(f"Sample sum: {float(sample_sum_exact):.2f}")
print(f"Prior shape: {prior_shape}")
print(f"Prior scale: {float(prior_scale_exact):.2f}")
print(f"Prior rate: {prior_rate_exact}")
```

## Likelihood

Conditional independence gives

$$
L(\theta;\mathbf{x})
=
\prod_{i=1}^{n}f(x_i\mid\theta).
$$

Substituting the sampling density,

$$
L(\theta;\mathbf{x})
=
\frac{\theta^{3n}}{\Gamma(3)^n}
\left(\prod_{i=1}^{n}x_i^2\right)
\exp\left(-\theta\sum_{i=1}^{n}x_i\right).
$$

The product of squared observations and the gamma-function term do not depend on $\theta$.

Consequently, as a function of $\theta$,

$$
L(\theta;\mathbf{x})
\propto
\theta^{3n}
\exp\left(-\theta\sum_{i=1}^{n}x_i\right).
$$

For the supplied sample summary,

$$
\sum_{i=1}^{n}x_i
=
n\overline{x}
=
10(64.22)
=
642.2.
$$

The individual observations are not needed to obtain the posterior kernel because their remaining contribution is independent of $\theta$.

## Posterior distribution

Bayes' rule gives

$$
\pi(\theta\mid\mathbf{x})
\propto
L(\theta;\mathbf{x})\pi(\theta).
$$

Multiplying the likelihood and prior kernels,

$$
\pi(\theta\mid\mathbf{x})
\propto
\theta^{3n}
e^{-\theta\sum_i x_i}
\theta^{\alpha_0-1}
e^{-\theta/\beta_0}.
$$

Collecting powers and exponential terms,

$$
\pi(\theta\mid\mathbf{x})
\propto
\theta^{\alpha_0+3n-1}
\exp\left[
-\theta\left(
\frac{1}{\beta_0}+\sum_{i=1}^{n}x_i
\right)
\right].
$$

Define the posterior shape and rate as

$$
a_n=\alpha_0+3n
$$

and

$$
r_n=\frac{1}{\beta_0}+\sum_{i=1}^{n}x_i.
$$

The normalized posterior density is

$$
\pi(\theta\mid\mathbf{x})
=
\frac{r_n^{a_n}}{\Gamma(a_n)}
\theta^{a_n-1}e^{-r_n\theta},
\qquad\theta>0.
$$

Therefore,

$$
\theta\mid\mathbf{x}
\sim
\operatorname{Gamma}
\left(\text{shape}=a_n,\ \text{scale}=\frac{1}{r_n}\right).
$$

For the supplied parameters,

$$
a_n=45+3(10)=75
$$

and

$$
r_n=\frac{1}{35}+642.2
=\frac{22478}{35}
\approx642.228571429.
$$

The posterior scale is

$$
\beta_n=\frac{1}{r_n}
=\frac{35}{22478}
\approx0.001557078032.
$$

Thus,

$$
\theta\mid\mathbf{x}
\sim
\operatorname{Gamma}
\left(
\text{shape}=75,\
\text{scale}=\frac{35}{22478}
\right)
$$

```python
posterior_shape = prior_shape + sample_shape * n

posterior_rate_exact = (
    prior_rate_exact + sample_sum_exact
)

posterior_scale_exact = 1 / posterior_rate_exact

posterior_rate = float(posterior_rate_exact)
posterior_scale = float(posterior_scale_exact)

assert posterior_shape > 0
assert posterior_rate_exact > 0
assert posterior_rate_exact * posterior_scale_exact == 1

print(f"Posterior shape: {posterior_shape}")
print(f"Exact posterior rate: {posterior_rate_exact}")
print(f"Posterior rate: {posterior_rate:.12f}")
print(f"Exact posterior scale: {posterior_scale_exact}")
print(f"Posterior scale: {posterior_scale:.12f}")
```

## Bayes estimate under squared-error loss

Let d be a proposed point estimate of $\theta$.

Under squared-error loss,

$$
L(\theta,d)=(\theta-d)^2.
$$

The posterior expected loss can be written as

$$
E[(\theta-d)^2\mid\mathbf{x}]
=
\operatorname{Var}(\theta\mid\mathbf{x})
+
\left(E[\theta\mid\mathbf{x}]-d\right)^2
$$

The variance term does not depend on d. The second term is minimized
when

$$
d=E[\theta\mid\mathbf{x}].
$$

Thus, the Bayes estimate under squared-error loss is the posterior mean.

For the posterior gamma density,

$$
E[\theta\mid\mathbf{x}]
=
\frac{r_n^{a_n}}{\Gamma(a_n)}
\int_0^\infty\theta^{a_n}e^{-r_n\theta}\,d\theta.
$$

Substituting $u=r_n\theta$ and using
$\Gamma(a_n+1)=a_n\Gamma(a_n)$ gives

$$
E[\theta\mid\mathbf{x}]=\frac{a_n}{r_n}
$$

Therefore,

$$
\widehat{\theta}_{SE}
=
\frac{75}{22478/35}
=
\frac{2625}{22478}.
$$

Numerically,

$$
\widehat{\theta}_{SE}\approx0.116780852389
$$

## Estimate based on the posterior mode

The posterior-mode estimate maximizes the posterior density.

Ignoring terms independent of $\theta$, the log-posterior is

$$
\log\pi(\theta\mid\mathbf{x})
=
C+(a_n-1)\log\theta-r_n\theta.
$$

Its derivative is

$$
\frac{d}{d\theta}\log\pi(\theta\mid\mathbf{x})
=
\frac{a_n-1}{\theta}-r_n.
$$

Setting the derivative equal to zero gives

$$
\widehat{\theta}_{MAP}
=
\frac{a_n-1}{r_n}.
$$

Here $a_n=75>1$, so the candidate lies inside the domain $\theta>0$.

Moreover,

$$
\frac{d^2}{d\theta^2}\log\pi(\theta\mid\mathbf{x})
=
-\frac{a_n-1}{\theta^2}<0.
$$

The log-posterior is strictly concave, so this stationary point is the unique global maximum.

Consequently,

$$
\widehat{\theta}_{MAP}
=
\frac{74}{22478/35}
=
\frac{2590}{22478}.
$$

Numerically,

$$
\widehat{\theta}_{MAP}\approx0.115223774357
$$

This is the posterior-mode estimate. It is distinct from the squared-error-loss estimate derived in part (b).

```python
posterior_mean_exact = (
    posterior_shape / posterior_rate_exact
)

assert posterior_shape > 1

posterior_mode_exact = (
    (posterior_shape - 1) / posterior_rate_exact
)

posterior_mean = float(posterior_mean_exact)
posterior_mode = float(posterior_mode_exact)

# stationary point and curvature of the log-posterior
mode_derivative = (
    (posterior_shape - 1) / posterior_mode
    - posterior_rate
)

mode_second_derivative = (
    -(posterior_shape - 1) / posterior_mode**2
)

assert abs(mode_derivative) < 1e-10
assert mode_second_derivative < 0

# here: mean minus mode = scale
assert (
    posterior_mean_exact - posterior_mode_exact
    == posterior_scale_exact
)

print(f"Squared-error Bayes estimate: {posterior_mean:.12f}")
print(f"Posterior-mode estimate: {posterior_mode:.12f}")
print(
    f"Mean minus mode: "
    f"{float(posterior_mean_exact - posterior_mode_exact):.12f}"
)
# Posterior estimate checks passed
```

## Interpretation

Under the stated shape-scale convention and sample size of ten, the posterior distribution of $\theta$ is gamma with shape 75 and scale approximately 0.001557078032.

The estimate minimizing posterior expected squared error is the posterior mean, approximately 0.116780852389.

The posterior density is maximized at approximately 0.115223774357. This posterior-mode estimate is slightly smaller than the mean, consistent with the right-skewed gamma posterior.

The estimates summarize the same posterior in different ways: one minimizes expected squared error, while the other identifies the location of maximum posterior density.

These results depend on the supplied prior and sampling model. They estimate the parameter $\theta$, which is the rate of the conditional sampling distribution, rather than the mean of the observations.
