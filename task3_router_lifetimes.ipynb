# Task 3: Router Bandwidth to Failure and Maximum Likelihood

## Given info

The assigned branch is $\xi_9=2$.

Let $S$ denote the total bandwidth processed by one router before its first hardware failure, measured in terabytes.

Its probability density is

$$
f_S(s;\theta)
=
\frac{s^6e^{-s/\theta}}{720\theta^7},
\qquad s\geq0,\quad\theta>0,
$$

and zero otherwise.

This is a gamma distribution with shape 7 and scale $\theta$,since $\Gamma(7)=6!=720$.

The observed bandwidth totals for five two-router systems are

$$
t_1=17,\quad t_2=64,\quad t_3=159,\quad t_4=53,\quad t_5=39
$$

in terabytes.

## Objectives

1. Express the combined system total in terms of the individual routers
2. State the modelling assumptions
3. Derive the probability density of the combined total
4. Construct the likelihood and a transformation preserving its maximum
5. Derive and calculate the maximum-likelihood estimate of $\theta$
6. Estimate the expected bandwidth total of the two-router system

```python
import math

# observations of the combined two-router system, in terabytes.
observations = [17.0, 64.0, 159.0, 53.0, 39.0]

single_router_shape = 7
number_of_routers = 2
system_shape = number_of_routers * single_router_shape

n = len(observations)
total_observed = sum(observations)
sample_mean = total_observed / n

assert n > 0
assert all(value > 0 for value in observations)

print(f"Number of observed systems: {n}")
print(f"Sum of observed totals: {total_observed:.1f} TB")
print(f"Sample mean: {sample_mean:.1f} TB")
```

## Combined system and assumptions

Let $S_1$ and $S_2$ denote the bandwidth totals to failure of the first and replacement routers.

The second router begins operating when the first router fails. Consequently, the combined bandwidth total is

$$
T=S_1+S_2.
$$

The model uses the following assumptions:

1. Both routers follow the supplied density with the same parameter $\theta$, reflecting comparable hardware and operating conditions.

2. The individual totals $S_1$ and $S_2$ are independent. In particular, the model excludes shared failure mechanisms that would create dependence between them.

3. The replacement router is initially functioning and does not suffer material degradation while waiting to be activated.

4. Switching to the replacement router does not introduce an additional failure mechanism or alter its bandwidth-to-failure distribution.

5. The observed system totals $T_1,\ldots,T_n$ are independent and identically distributed, with a common parameter $\theta$.

6. All observations represent complete system totals to failure. The observations are not censored, and measurement error is neglected.


## Density of the combined bandwidth total

Under independence, the joint density of $S_1$ and $S_2$ is the product of their individual densities.

Using the transformation $T=S_1+S_2$ and $U=S_1$, the inverse transformation is

$$
S_1=U,\qquad S_2=T-U.
$$

The absolute Jacobian determinant is one. For a fixed $t>0$, the possible values of $U$ satisfy $0<u<t$.

Integrating out $U$ therefore gives

$$
f_T(t;\theta)
=
\int_0^t f_S(u;\theta)f_S(t-u;\theta)\,du.
$$

Substituting the supplied density,

$$
f_T(t;\theta)
=
\frac{e^{-t/\theta}}{720^2\theta^{14}}
\int_0^t u^6(t-u)^6\,du.
$$

Set $v=u/t$, so that $u=tv$ and $du=t\,dv$. Then

$$
\int_0^t u^6(t-u)^6\,du
=
t^{13}\int_0^1 v^6(1-v)^6\,dv.
$$

For positive integers $a,b$, repeated integration by parts gives

$$
\int_0^1 v^{a-1}(1-v)^{b-1}\,dv
=
\frac{(a-1)!(b-1)!}{(a+b-1)!}.
$$

To see this, denote the integral by $I(a,b)$. For $b>1$, integration by parts yields

$$
I(a,b)=\frac{b-1}{a}I(a+1,b-1),
$$

while $I(a,1)=1/a$. Repeating the recurrence gives the stated formula.

With $a=b=7$,

$$
\int_0^1 v^6(1-v)^6\,dv
=
\frac{6!\,6!}{13!}
=
\frac{720^2}{13!}.
$$

Consequently,

$$
f_T(t;\theta)
=
\frac{t^{13}e^{-t/\theta}}{13!\theta^{14}},
\qquad t\geq0,\quad\theta>0.
$$

The density is zero for $t<0$.

Thus, $T$ has a gamma distribution with shape 14 and scale $\theta$.

## Likelihood and log-likelihood

For independent observed system totals $t_1,\ldots,t_n$, the likelihood is the product of the individual densities:

$$
L(\theta;\mathbf{t})
=
\prod_{i=1}^{n}
\frac{t_i^{13}e^{-t_i/\theta}}{13!\theta^{14}},
\qquad \theta>0.
$$

Collecting the factors gives

$$
L(\theta;\mathbf{t})
=
\frac{\prod_{i=1}^{n}t_i^{13}}
{(13!)^n\theta^{14n}}
\exp\left(-\frac{\sum_{i=1}^{n}t_i}{\theta}\right).
$$

All observations are positive, so the likelihood is positive for every $\theta>0$.

The natural logarithm is strictly increasing. Therefore, maximizing the likelihood is equivalent to maximizing the log-likelihood:

$$
\ell(\theta)
=
13\sum_{i=1}^{n}\log t_i
-
n\log(13!)
-
14n\log\theta
-
\frac{\sum_{i=1}^{n}t_i}{\theta}.
$$

The logarithmic transformation turns products into sums and makes differentiation simpler.

The first two terms do not depend on $\theta$, so they do not affect the location of the maximum.

## Maximum-likelihood estimator

Differentiating the log-likelihood gives

$$
\ell'(\theta)
=
-\frac{14n}{\theta}
+
\frac{\sum_{i=1}^{n}t_i}{\theta^2}
=
\frac{\sum_{i=1}^{n}t_i-14n\theta}{\theta^2}.
$$

Setting the derivative equal to zero yields

$$
\widehat{\theta}
=
\frac{\sum_{i=1}^{n}t_i}{14n}
=
\frac{\overline{t}}{14}.
$$

Because $\theta^2>0$, the sign of the derivative is determined by its numerator:

$$
\ell'(\theta)>0
\quad\text{when}\quad
0<\theta<\widehat{\theta},
$$

and

$$
\ell'(\theta)<0
\quad\text{when}\quad
\theta>\widehat{\theta}.
$$

Thus, the log-likelihood increases up to $\widehat{\theta}$ and decreases afterwards. The estimator is the unique global maximizer on $\theta>0$.

For the supplied observations,

$$
\sum_{i=1}^{5}t_i
=
17+64+159+53+39
=
332.
$$

Therefore,

$$
\widehat{\theta}
=
\frac{332}{14\cdot5}
=
\frac{166}{35}
\approx4.742857\ \text{TB}.
$$

```python
theta_hat = total_observed / (system_shape * n)


def log_likelihood(theta):
    """log-likelihood for the observed combined-system totals"""
    if theta <= 0:
        return -math.inf

    return (
        (system_shape - 1)
        * sum(math.log(value) for value in observations)
        - n * math.lgamma(system_shape)
        - system_shape * n * math.log(theta)
        - total_observed / theta
    )


def score(theta):
    """derivative of the log-likelihood"""
    if theta <= 0:
        raise ValueError("theta must be positive")

    return (
        -system_shape * n / theta
        + total_observed / theta**2
    )


assert theta_hat > 0
assert abs(score(theta_hat)) < 1e-10

assert score(0.9 * theta_hat) > 0
assert score(1.1 * theta_hat) < 0

assert log_likelihood(theta_hat) > log_likelihood(0.9 * theta_hat)
assert log_likelihood(theta_hat) > log_likelihood(1.1 * theta_hat)

print(f"Estimated scale parameter: {theta_hat:.9f} TB")
print(f"Score at the estimate: {score(theta_hat):.3e}")
print(f"Log-likelihood at the estimate: {log_likelihood(theta_hat):.9f}")
```

## Expected bandwidth total of the two-router system

Using the derived system density,

$$
E[T]
=
\int_0^\infty
t\frac{t^{13}e^{-t/\theta}}{13!\theta^{14}}\,dt.
$$

With the substitution $u=t/\theta$,

$$
E[T]
=
\frac{\theta}{13!}
\int_0^\infty u^{14}e^{-u}\,du.
$$

Repeated integration by parts gives

$$
\int_0^\infty u^{14}e^{-u}\,du=14!.
$$

Hence,

$$
E[T]
=
\frac{\theta\,14!}{13!}
=
14\theta.
$$

Replacing the unknown parameter with its maximum-likelihood estimate gives

$$
\widehat{E[T]}
=
14\widehat{\theta}
=
14\frac{332}{70}.
$$

Therefore,

$$
\widehat{E[T]}=66.4\ \text{TB}
$$

The fitted expected system total equals the sample mean because
$\widehat{\theta}=\overline{t}/14$.

```python
estimated_system_mean = system_shape * theta_hat

assert math.isclose(
    estimated_system_mean,
    sample_mean,
    rel_tol=1e-12,
)

print(f"Estimated theta: {theta_hat:.6f} TB")
print(f"Estimated expected system total: {estimated_system_mean:.1f} TB")
print(f"Observed sample mean: {sample_mean:.1f} TB")
```

## Interpretation

Under the stated assumptions, the combined bandwidth total of two sequential routers follows a gamma distribution with shape 14 and scale $\theta$.

The five observed system totals give a maximum-likelihood scale estimate of approximately 4.742857 TB. The fitted expected total bandwidth processed before both routers fail is 66.4 TB.

This expectation describes an average across repeated systems operating under the model. It is not a guaranteed minimum or the predicted total for every individual system.

The estimate is based on five complete observations and depends on the assumptions of independent router totals, comparable operating conditions, and a common scale parameter. The calculations do not establish those assumptions empirically.
